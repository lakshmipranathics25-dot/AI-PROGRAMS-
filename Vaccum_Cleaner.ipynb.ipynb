```python
def vacuum_cleaner_agent(room_a, room_b, location):

    print("\nInitial State:")
    print("Room A:", room_a)
    print("Room B:", room_b)
    print("Vacuum Location:", location)

    while room_a == "Dirty" or room_b == "Dirty":

        if location == "A":
            if room_a == "Dirty":
                print("\nVacuum is in Room A.")
                print("Room A is Dirty -> Cleaning Room A...")
                room_a = "Clean"
                print("Room A is now Clean.")

            if room_b == "Dirty":
                print("Moving from Room A to Room B...")
                location = "B"

        elif location == "B":
            if room_b == "Dirty":
                print("\nVacuum is in Room B.")
                print("Room B is Dirty -> Cleaning Room B...")
                room_b = "Clean"
                print("Room B is now Clean.")

            if room_a == "Dirty":
                print("Moving from Room B to Room A...")
                location = "A"

    print("\nFinal State:")
    print("Room A:", room_a)
    print("Room B:", room_b)
    print("Vacuum Location:", location)
    print("\nGoal achieved! Both rooms are clean.")


print("     VACUUM CLEANER AGENT")

room_a = input("Enter status of Room A (Clean/Dirty): ").capitalize()
room_b = input("Enter status of Room B (Clean/Dirty): ").capitalize()
location = input("Enter vacuum location (A/B): ").upper()

if room_a not in ["Clean", "Dirty"]:
    print("Invalid status for Room A.")
elif room_b not in ["Clean", "Dirty"]:
    print("Invalid status for Room B.")
elif location not in ["A", "B"]:
    print("Invalid vacuum location.")
else:
    vacuum_cleaner_agent(room_a, room_b, location)
```
